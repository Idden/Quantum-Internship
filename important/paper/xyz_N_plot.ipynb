```python
import numpy as np
import matplotlib.pyplot as plt

xdata4 = np.load("xyz_data/x_dis_N4.npz")
ydata4 = np.load("xyz_data/y_dis_N4.npz")
zdata4 = np.load("xyz_data/z_dis_N4.npz")

xdata8 = np.load("xyz_data/x_dis_N8.npz")
ydata8 = np.load("xyz_data/y_dis_N8.npz")
zdata8 = np.load("xyz_data/z_dis_N8.npz")

xdata14 = np.load("xyz_data/x_dis_N14.npz")
ydata14 = np.load("xyz_data/y_dis_N14.npz")
zdata14 = np.load("xyz_data/z_dis_N14.npz")
```

```python
tlist = xdata4["tlist"]

xscar4 = xdata4["scar"]
yscar4 = ydata4["scar"]
zscar4 = zdata4["scar"]

xscar8 = xdata8["scar"]
yscar8 = ydata8["scar"]
zscar8 = zdata8["scar"]

xscar14 = xdata14["scar"]
yscar14 = ydata14["scar"]
zscar14 = zdata14["scar"]
```

```python
fig, axs = plt.subplots(3, 1, figsize=(6, 8), sharex=True)
fig.suptitle(r"$\sigma_{xyz}$ Disorder Comparisons")

axs[0].plot(tlist, xscar4, label="4 Qubits")
axs[0].plot(tlist, xscar8, label="8 Qubits")
axs[0].plot(tlist, xscar14, label="14 Qubits")
axs[0].set_title("X Disorder")
axs[0].set_ylabel(r"$R(\tau)$")
axs[0].legend()

axs[1].plot(tlist, yscar4, label="4 Qubits")
axs[1].plot(tlist, yscar8, label="8 Qubits")
axs[1].plot(tlist, yscar14, label="14 Qubits")
axs[1].set_title("Y Disorder")
axs[1].set_ylabel(r"$R(\tau)$")
axs[1].legend()

axs[2].plot(tlist, zscar4, label="4 Qubits")
axs[2].plot(tlist, zscar8, label="8 Qubits")
axs[2].plot(tlist, zscar14, label="14 Qubits")
axs[2].set_title("Z Disorder")
axs[2].set_xlabel("Time")
axs[2].set_ylabel(r"$R(\tau)$")
axs[2].legend()

axs[0].set_ylim(0, 1)
axs[1].set_ylim(0, 1)
axs[2].set_ylim(0, 1)

plt.savefig(f"figures/xyz_N_4_8_14_dis{0.3}_reals{500}.pdf")
plt.tight_layout()
plt.show()
```
