```python
import numpy as np
import qutip as qt
import matplotlib.pyplot as plt
from helper.quantumScarFunctions import *
from helper.quantumScarsPlotting import *
import time
```

```python
N = 6
tau_r = get_tau_r(N)
wd = 2*np.pi / tau_r
wm = 1.0
t_max = 200
tlist = np.linspace(0, t_max, 5000)
reals = 1
rand = True
dis = 0.0
```

```python
dz = 0.0
dy = 0.0
dx = dis

args = {"A": 0.1, "omega": wd}
qargs = {"A": 0.1, "omega": wm}

H0_clean, eigenvalues, eigenstates, psi0, basisList = get_scar_ham(N)
H1, driveWeights = get_scar_H1(N, basisList)

t0 = time.perf_counter()
full_scar = 0
full_overlap = 0
for _ in range(reals):
    H0, eigenvalues, eigenstates = get_dis_scar_ham(H0_clean, N, basisList, ham_disorder=[dz, dy, dx])
    bandwidth = eigenvalues[-1] - eigenvalues[0]

    H = qt.QobjEvo([H0, [H1, coeff]], args=args)
    psi_t = qt.sesolve(H, eigenstates[0], tlist, e_ops=[H0], options={"store_states": True})
    scarIndices, scarStates = get_scar_states(N, psi0)
    scarProbs, individualProbs = scar_overlap_from_states(N, tlist, scarStates, psi_t, save_fig=False, plot_fig=False, indiv=True)
    Rtau_scar = np.array(np.real(psi_t.expect[0] - psi_t.expect[0][0]) / bandwidth)
    full_scar += Rtau_scar

    P = np.asarray(scarProbs)
    if P.ndim == 2:
        P = P.sum(axis=1) if P.shape[0] == len(tlist) else P.sum(axis=0)
    full_overlap += P

xfull_scar = full_scar / reals
xfull_overlap = full_overlap / reals

full_qubit = 0
for _ in range(reals):
    qH0_list, qH1_list, _ = get_qubit_ham(N, wm=wm, ham_disorder=[dz, dy, dx])

    dE_tot = 0
    bw_tot = 0
    for qH0, qH1 in zip(qH0_list, qH1_list):
        qeigenvalues, qeigenstates = qH0.eigenstates()
        bw_tot += qeigenvalues[-1] - qeigenvalues[0]

        qH = qt.QobjEvo([qH0, [qH1, coeff]], args=qargs)
        qpsi_t = qt.sesolve(qH, qeigenstates[0], tlist, e_ops=[qH0])
        dE_tot = dE_tot + np.real(qpsi_t.expect[0] - qpsi_t.expect[0][0])

    full_qubit += dE_tot / bw_tot

xfull_qubit = full_qubit / reals
elapsed = time.perf_counter() - t0

print(f"{reals} realizations in {elapsed:.1f} s "
      f"({elapsed/reals:.5f} s per realization)")

fig, (ax_a, ax_b) = plt.subplots(2, 1, sharex=True, figsize=(6, 7.5))

ax_a.plot(tlist, xfull_scar, "-", label="Scar")
ax_a.plot(tlist, xfull_qubit, "--", label="Decoupled")
ax_a.set_ylabel(r"$R(\tau)$")
ax_a.set_ylim(0, 1)
ax_a.legend(loc="upper right")
ax_a.text(0.02, 0.90, "(a)", transform=ax_a.transAxes)

ax_b.plot(tlist, xfull_overlap, "-")
ax_b.set_ylabel("Scar subspace weight")
ax_b.set_xlabel(r"$\tau$")
ax_b.set_ylim(0, 1.05)
ax_b.text(0.02, 0.90, "(b)", transform=ax_b.transAxes)

fig.tight_layout()
fig.savefig(f"Rtau_overlap_N{N}.pdf")
plt.show()

for i in range(len(individualProbs)):
    plt.plot(tlist, individualProbs[i], label=f"Scar {i+1}")
    plt.xlabel("Time")
    plt.ylabel("Individual Scar Probability")
    plt.title("Individual Scar Probabilities vs. Time")
    plt.ylim(0, 1)
    plt.legend()
    plt.show()
```
